### 1.Dataset

A dataset is a structured collection of data containing multiple observations or records and their corresponding characteristics.

Real-world example:

A telecom company may maintain customer information such as:

* Age
* Contract type
* Monthly charges
* Internet service
* Payment method
* Customer churn

Each customer represents one record, while each characteristic represents a feature.

```python
df.head(10)
```

### 2.Feature

A feature is an individual variable or column that describes some characteristic of an observation.

For example:

* Age
* MonthlyCharges
* ContractType
* InternetService
* PaymentMethod

These are all features of a customer.

Real-world example:

For predicting whether a customer will leave a telecom company:

* Age → feature
* Monthly charges → feature
* Contract type → feature
* Internet service → feature

```python
features = df.columns

print(features)
```

### 3.Target

The target is the variable that we want to predict or explain.

In our dataset:

Churn is the target.

Real-world example:

A telecom company wants to predict:

"Will this customer leave the company?"

Therefore:

Target = Churn

```python
target = "Churn"

print(df[target].head())
```

### 10. Identifier Columns

An identifier column uniquely identifies each record.

In our dataset:

CustomerID is an identifier.

Example:

CUST00001

CUST00002

CUST00003

#### Important ML concept:

An identifier generally should not be used as a predictive feature because it doesn't represent meaningful customer behavior.

For example:

CustomerID = CUST00001

doesn't tell us whether the customer will churn.

```python
print(df["CustomerID"].head())
```
